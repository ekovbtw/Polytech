```python
import pandas as pd
import random
import matplotlib.pyplot as plt

# Настройка, чтобы графики отображались прямо в Jupyter
%matplotlib inline
random.seed(42)

with open('var_9_dgamma.csv', 'r', encoding='utf-8') as file:
    X = [float(line.strip()) for line in file if line.strip() != '']

n = len(X)
print(f"Объём выборки: n = {n}")
print(f"Первые 10 элементов выборки: {X[:10]}")

```

```python
def sample_sum(X):
    """
    Вычисляет сумму элементов выборки: S = x1 + x2 + ... + xn
    """
    s = 0.0
    for x in X:
        s += x
    return s
```

```python
def sample_mean(X):
    """
    Вычисляет выборочное среднее: x_ = (1/n) * sum(xi)
    """
    return sample_sum(X) / len(X)
```

```python
def sample_median(X):
    """
    Вычисляет медиану выборки — значение, делящее упорядоченную выборку пополам.
    При чётном n медиана равна среднему двух центральных элементов.
    """
    X_sorted = sorted(X)          # только упорядочивание, не статистическая функция
    n = len(X_sorted)
    mid = n // 2
    if n % 2 == 1:
        return X_sorted[mid]
    else:
        return (X_sorted[mid - 1] + X_sorted[mid]) / 2
```

```python
def sample_mode(X):
    """
    Вычисляет моду выборки — значение, встречающееся чаще всего.
    Если все значения выборки уникальны (что типично для непрерывной СВ),
    все частоты равны 1, и функция вернёт первое встретившееся значение
    выборки (устойчивое поведение при равных частотах).
    """
    freq = {}
    for x in X:
        freq[x] = freq.get(x, 0) + 1

    best_value, best_count = None, 0
    for value, count in freq.items():
        if count > best_count:
            best_count = count
            best_value = value
    return best_value
```

```python
def sample_range(X):
    """
    Вычисляет размах выборки: R = x_max - x_min
    """
    x_min = x_max = X[0]
    for x in X:
        if x < x_min:
            x_min = x
        if x > x_max:
            x_max = x
    return x_max - x_min
```

```python
def sample_variance_biased(X):
    """
    Вычисляет смещённую (выборочную) дисперсию: D = (1/n) * sum((xi - x_)^2)
    """
    mean = sample_mean(X)
    n = len(X)
    s = 0.0
    for x in X:
        s += (x - mean) ** 2
    return s / n
```

```python
def sample_variance_unbiased(X):
    """
    Вычисляет несмещённую дисперсию: S^2 = (1/(n-1)) * sum((xi - x_)^2)
    """
    mean = sample_mean(X)
    n = len(X)
    s = 0.0
    for x in X:
        s += (x - mean) ** 2
    return s / (n - 1)
```

```python
def initial_moment(X, k):
    """
    Вычисляет выборочный начальный момент k-го порядка:
    m_k = (1/n) * sum(xi^k)
    """
    n = len(X)
    s = 0.0
    for x in X:
        s += x ** k
    return s / n
```

```python
def central_moment(X, k):
    """
    Вычисляет выборочный центральный момент k-го порядка:
    mu_k = (1/n) * sum((xi - x_)^k)
    """
    mean = sample_mean(X)
    n = len(X)
    s = 0.0
    for x in X:
        s += (x - mean) ** k
    return s / n
```

```python
k = 3  # порядок момента (по заданию — любой, кроме 1 и 2)

print(f"Количество элементов выборки:        n = {n}")
print(f"Сумма элементов выборки:             {sample_sum(X):.6f}")
print(f"Выборочное среднее:                  {sample_mean(X):.6f}")
print(f"Медиана:                             {sample_median(X):.6f}")
print(f"Мода:                                {sample_mode(X):.6f}")
print(f"Размах выборки:                      {sample_range(X):.6f}")
print(f"Смещённая дисперсия:                 {sample_variance_biased(X):.6f}")
print(f"Несмещённая дисперсия:               {sample_variance_unbiased(X):.6f}")
print(f"Начальный момент {k}-го порядка:        {initial_moment(X, k):.6f}")
print(f"Центральный момент {k}-го порядка:      {central_moment(X, k):.6f}")
```
